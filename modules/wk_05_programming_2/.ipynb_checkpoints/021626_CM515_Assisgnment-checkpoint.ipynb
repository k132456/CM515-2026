## 1.Create a function called `average()` that takes a single argument (a list of numbers) and returns the mean as part of an f-string.

```python
def average(list_of_nums):
    sum = 0
    for items in list_of_nums:
        sum = sum + items
    #print(sum)
    mean = sum/(len(list_of_nums))
    return f"The mean is {mean}."

listA = [3,3,3,3,3,3,3,3]
average(listA)

listB = [5,7,2,3,45]
average(listB)
```

## 2. Make a function called `revcomp()` that prompts the user for a DNA sequence and returns the reverse complement. Have the function inform the user if they put RNA instead.

```python
def revcomp():
    output = ""
    question = input("Please provide the DNA sequence.")
    if "U" in question:
        print ("You have entered an RNA strand instead of a DNA strand.")
    else:
        for items in question:
            if items == "A":
                output = output + "T"
            elif items == "T":
                output = output + "A"
            elif items == "G":
                output = output + "C"
            elif items == "C":
                output = output + "G"
        return output



revcomp()
```

```python
def savings_calc():
    principle = input("What was the principle amount of money that you started the account with?")
    principle = int(principle)
    
    interest = input("What is the annual interest rate?")
    interest = float(interest)
    
    years = input("How long has the account been open, in years?")
    years = float(years)
    
    final_savings = round((principle * (((1) + (interest/1)) ** (1 * years))), 2)
    
    return f"You started with ${principle} in your savings account which has {interest} annual interest. You ended up with ${final_savings} after investing for {years} years."
savings_calc()
```

# 6. Try making a function called `average()`.

This function should:

    Prompt the user for input continuously.
        
    Be able to provide the user with an average of numbers when they are done providing numbers (by having the user type someting like 'stop', or 'finished').

    Give the user an option to exit out of the program.
        
    Be able to address common pitfalls a user might have (There are two big ones).


```python
def average():
    #question = input("Please enter the numbers, one by one, that you would like to find the average of. When you are done entering numbers, enter STOP.")
    #while question = <class 'int'>:
    numbers_list = []
    while True:
        question = input("Please enter the numbers that you would like to find the average of, one by one. When you are done entering numbers, enter 'STOP'.")
        if question == "STOP":
            break
        numbers_list.append(float(question))
    average_of_num = ((sum(numbers_list))/(len(numbers_list)))
    return average_of_num

average()

        
```
